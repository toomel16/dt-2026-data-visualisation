```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
df.describe() # Show basic statistics
```

```python
import matplotlib.pyplot as plt # matplotlib can create charts and data visualisations
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of Practice per Child")
plt.ylabel("Years of Practice")
plt.show()
```

```python
import matplotlib.pyplot as plt # matplotlib can create charts and data visualisations
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age of Athlete")
plt.ylabel("Age")
plt.show()
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Share of Age from Sum")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.6) 
# s=df["Years_of_practice"]*60 means that if the student has more years of practice the bubble will appear larger
# alpha=0.6 makes the bubbles transparent
plt.ylabel("Years of Practice")
plt.xlabel("Age")
plt.title("Years of Practice vs. Age")
```

```python
import seaborn as sns # used for cleaner and better formatted boxplots
sns.boxplot(data=df, y="Age")
plt.title("Distribution of Age")
plt
```

```python
plt.Figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.show
```

```python
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
```

```python
avg_yrs_prc= df.groupby("Sport")["Years_of_practice"].mean()
avg_yrs_prc.plot(kind="bar")
```
